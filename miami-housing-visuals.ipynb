```python
from pathlib import Path
import pandas as pd

base_dir = Path.cwd()
miami_data_path = base_dir / "archive (1)" / "miami-housing.csv"

if not miami_data_path.exists():
    miami_data_path = base_dir.parent / "archive (1)" / "miami-housing.csv"

if not miami_data_path.exists():
    raise FileNotFoundError(f"CSV not found: {miami_data_path}")

miami_data = pd.read_csv(miami_data_path)
miami_data.info()
```

```python
import matplotlib.pyplot as plt
plt.style.use('classic')
import seaborn as sns

# Plot the distribution of Miami housing prices, left skewed thus most prices are concentrated on the lower end
fig, ax = plt.subplots()
sns.histplot(miami_data['SALE_PRC'], kde=True, ax=ax)
ax.set_title('Distribution of Miami Housing Prices')
ax.set_xlabel('Price')
ax.set_ylabel('Frequency')
plt.show()
```

```python
import plotly.express as px

#Put houses in a price range and assort by color 
miami_data['PRICE_RANGE'] = pd.cut(miami_data['SALE_PRC'], bins=[0, 200000, 400000, 600000, 800000, 1000000, 2000000, 5000000], labels=['0-200k', '200k-400k', '400k-600k', '600k-800k', '800k-1M', '1M-2M', '2M-5M'])

#Plot the longitude and lattitude on the map using a color to represent the sq ft
fig = px.scatter_mapbox(miami_data,
                        lat="LATITUDE",
                        lon="LONGITUDE",
                        size="LND_SQFOOT",
                        color="PRICE_RANGE",
                        color_discrete_sequence=px.colors.qualitative.Plotly,
                        size_max=18,
                        zoom=10,
                        hover_name="SALE_PRC",
                        hover_data = ["LND_SQFOOT", "PRICE_RANGE"]
                        )

fig.update_layout(mapbox_style="open-street-map")
fig.update_layout(margin={"r":0,"t":0,"l":0,"b":0})
fig.show()
```

```python
# Use a log-scaled y-axis and bubble sizing so the expensive homes stand out without flattening the smaller ones
fig2 = px.scatter(
    miami_data,
    x="LND_SQFOOT",
    y="SALE_PRC",
    color="PRICE_RANGE",
    size="LND_SQFOOT",
    size_max=18,
    opacity=0.7,
    log_y=True,
    color_discrete_sequence=px.colors.qualitative.Plotly,
    hover_name="SALE_PRC",
    hover_data=["LND_SQFOOT", "PRICE_RANGE"],
    title="Miami Home Price vs Lot Size (log-scaled)"
)
fig2.update_layout(
    template="plotly_white",
    xaxis_title="Lot Size (sq ft)",
    yaxis_title="Sale Price",
    legend_title="Price Range",
    margin={"r": 10, "t": 40, "l": 10, "b": 10}
)
fig2.show()
```

```python
#Comparing the features of water distance to the price to see if there is a relationship. 
import seaborn as sns

miami_data[["WATER_DIST", "SALE_PRC"]].corr()

#Visualizing the relationship between water distance and sale price by using a heat map 
sns.heatmap(miami_data[["WATER_DIST", "SALE_PRC"]].corr(), annot=True, cmap="coolwarm")

#The heat map shows that there is a weak negative correlation between water distance and sale price, indicating that properties closer to water tend to have slightly higher prices.
```

```python
#add on scatter/regresion
import statsmodels.api as sm

x = miami_data["WATER_DIST"]
y = miami_data["SALE_PRC"]
x = sm.add_constant(x)
model = sm.OLS(y, x).fit()
print(model.summary())

fig4 = px.scatter(
    miami_data,
    x="WATER_DIST",
    y="SALE_PRC",
    trendline="ols",
    trendline_color_override="red",
    title = "Water Distance vs Sale Price with Linear Regression",
    labels = {
        "WATER_DIST": "Distance from Water",
        "SALE_PRC": "Sale Price"
    }, 
    hover_data=["WATER_DIST", "SALE_PRC"],
    opacity=0.7,
)
fig4.show()


#sns.scatterplot(x="WATER_DIST", y="SALE_PRC", data=miami_data); not useful too many observations 
```

```python
#Add a linear regressions to test the relationships between variables and the sale price.
#lot size vs sale price
x = miami_data["LND_SQFOOT"]
y = miami_data["SALE_PRC"]
x = sm.add_constant(x)
model = sm.OLS(y, x).fit()
print(model.summary())

fig5 = px.scatter(
    miami_data,
    x="LND_SQFOOT",
    y="SALE_PRC",
    trendline="ols",
    trendline_color_override="red",
    title = "Lot Size vs Sale Price with Linear Regression",
    labels = {
        "LND_SQFOOT": "Lot Size (Square Foot)",
        "SALE_PRC": "Sale Price"
    }, 
    hover_data=["LND_SQFOOT", "SALE_PRC"],
    opacity=0.7,
)
fig5.show()
#the larger the lot size, the higher the sale price tends to be
```

```python
#ocean distance vs sale price
x = miami_data["OCEAN_DIST"]
y = miami_data["SALE_PRC"]
x = sm.add_constant(x)
model = sm.OLS(y, x).fit()
print(model.summary())

fig6 = px.scatter(
    miami_data,
    x="OCEAN_DIST",
    y="SALE_PRC",
    trendline="ols",
    trendline_color_override="red",
    title = "Ocean Distance vs Sale Price with Linear Regression",
    labels = {
        "OCEAN_DIST": "Distance from Ocean",
        "SALE_PRC": "Sale Price"
    }, 
    hover_data=["OCEAN_DIST", "SALE_PRC"],
    opacity=0.7,
)
fig6.show()
#the farther the property is from the ocean, the lower the sale price tends to be
```

```python
#special feature valuse vs sale price
x = miami_data["SPEC_FEAT_VAL"]
y = miami_data["SALE_PRC"]
x = sm.add_constant(x)
model = sm.OLS(y, x).fit()
print(model.summary())

fig7 = px.scatter(
    miami_data,
    x="SPEC_FEAT_VAL",
    y="SALE_PRC",
    trendline="ols",
    trendline_color_override="red",
    title = "Special Feature Values vs Sale Price with Linear Regression",
    labels = {
        "SPEC_FEAT_VAL": "Special Feature Values",
        "SALE_PRC": "Sale Price"
    }, 
    hover_data=["SPEC_FEAT_VAL", "SALE_PRC"],
    opacity=0.7,
)
fig7.show()
#the more special features a property has, the higher the sale price tends to be
```

```python
#Age vs Sale price 
x = miami_data["age"]
y = miami_data["SALE_PRC"]
x = sm.add_constant(x)
model = sm.OLS(y, x).fit()
print(model.summary())

fig8 = px.scatter(
    miami_data,
    x="age",
    y="SALE_PRC",
    trendline="ols",
    trendline_color_override="red",
    title = "Age vs Sale Price with Linear Regression",
    labels = {
        "age": "Age",
        "SALE_PRC": "Sale Price"
    }, 
    hover_data=["age", "SALE_PRC"],
    opacity=0.7,
)
fig8.show()

#older the property, the lower the sale price tends to be
```
